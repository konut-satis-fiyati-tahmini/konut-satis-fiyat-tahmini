```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Read the data
df = pd.read_csv('train.csv')
initial_shape = df.shape

# 2. Scatter plot before cleaning
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.scatterplot(x=df['GrLivArea'], y=df['SalePrice'], alpha=0.6)
plt.title("Temizlik Öncesi (Aykırı Değerler)")

# 3. Clean Outliers
df_cleaned = df.drop(df[(df['GrLivArea'] > 4000)].index)
cleaned_shape = df_cleaned.shape

# Scatter plot after cleaning
plt.subplot(1, 2, 2)
sns.scatterplot(x=df_cleaned['GrLivArea'], y=df_cleaned['SalePrice'], alpha=0.6)
plt.title("Temizlik Sonrası")
plt.tight_layout()
plt.savefig('scatter_outliers.png')
plt.close()

# 4. Fill Missing Values on df_cleaned
cols_none = ['PoolQC', 'MiscFeature', 'Alley', 'Fence', 'FireplaceQu', 
             'GarageType', 'GarageFinish', 'GarageQual', 'GarageCond', 
             'BsmtQual', 'BsmtCond', 'BsmtExposure', 'BsmtFinType1', 'BsmtFinType2']
for col in cols_none:
    if col in df_cleaned.columns:
        df_cleaned[col] = df_cleaned[col].fillna('None')

cols_zero = ['GarageYrBlt', 'GarageArea', 'GarageCars', 
             'BsmtFinSF1', 'BsmtFinSF2', 'BsmtUnfSF','TotalBsmtSF', 
             'BsmtFullBath', 'BsmtHalfBath', 'MasVnrArea']
for col in cols_zero:
    if col in df_cleaned.columns:
        df_cleaned[col] = df_cleaned[col].fillna(0)
# LotFrontage eksiklerini, evin bulunduğu mahallenin (Neighborhood) medyan değeriyle doldurma
if 'Neighborhood' in df_cleaned.columns and 'LotFrontage' in df_cleaned.columns:
    df_cleaned['LotFrontage'] = df_cleaned.groupby('Neighborhood')['LotFrontage'].transform(lambda x: x.fillna(x.median()))

if 'MasVnrType' in df_cleaned.columns:
    df_cleaned['MasVnrType'] = df_cleaned['MasVnrType'].fillna('None')
if 'Electrical' in df_cleaned.columns:
    df_cleaned['Electrical'] = df_cleaned['Electrical'].fillna(df_cleaned['Electrical'].mode()[0])

missing_after = df_cleaned.isnull().sum()[df_cleaned.isnull().sum() > 0]

# 5. Correlation Heatmap
numeric_df = df_cleaned.select_dtypes(include=[np.number])
corr_matrix = numeric_df.corr()
if 'SalePrice' in corr_matrix.columns:
    top_corr_features = corr_matrix.index[abs(corr_matrix["SalePrice"]) > 0.5]
    plt.figure(figsize=(10, 8))
    sns.heatmap(df_cleaned[top_corr_features].corr(), annot=True, cmap="coolwarm", linewidths=.5, fmt=".2f")
    plt.title("Satış Fiyatı ile Yüksek Korelasyonlu Özellikler")
    plt.tight_layout()
    plt.savefig('heatmap.png')
    plt.close()
    
print(f"Başlangıç boyutu: {initial_shape}")
print(f"Temizlik sonrası boyut: {cleaned_shape}")
if len(missing_after) > 0:
    print(f"Kalan eksik veriler:\n{missing_after}")
else:
    print("Tüm eksik veriler temizlendi.")
```
